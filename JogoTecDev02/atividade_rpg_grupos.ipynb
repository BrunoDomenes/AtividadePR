# ⚔️ O Grande Jogo: Desenvolvimento Colaborativo Open Source

Chegou a hora de criarmos o nosso próprio universo de RPG! Vocês foram divididos em 3 grandes estúdios de desenvolvimento. Cada grupo cuidará de um pilar do nosso jogo.

**A Regra de Ouro:** Ninguém mexe na pasta do outro grupo. O código principal está com o Mestre (Professor), e vocês usarão o fluxo de **Fork e Pull Request** para enviar as suas criações!

## 🚀 O Fluxo de Trabalho (Para todos os grupos)

1. **Fork:** Façam o Fork do repositório do Mestre no GitHub de vocês.
2. **Clone:** Abram o VS Code e clonem o Fork de vocês (`git clone URL`).
3. **Criação:** Entrem na pasta do seu grupo e criem os arquivos `.md` conforme as instruções abaixo.
4. **Envio Local:** `git add .` ➔ `git commit -m "mensagem"` ➔ `git push origin main`.
5. **Pull Request:** Voltem no GitHub e abram o PR para enviar o código para o Mestre!
6. **Sync:** Quando o Mestre aprovar todos, façam o *Sync Fork* no GitHub e rodem `git pull origin main` no VS Code para verem o jogo completo.

## 🧙‍♂️ Grupo 1: Personagens
**Pasta de Trabalho:** `/personagens`

Vocês são responsáveis por criar as Classes jogáveis. Cada aluno do grupo deve criar **1 arquivo** para uma classe diferente (ex: `mago.md`, `guerreiro.md`, `hacker.md`).

**💡 Dica de Estrutura para o Arquivo:**
```markdown
# Nome da Classe (Ex: Arqueiro Cibernético)

**Descrição:** Breve resumo de quem é e de onde vem.

### Status Base
- 💖 Vida (HP): 100
- ⚡ Energia/Mana: 50

### Atributos
- 💪 Força: 2
- 🏃‍♂️ Agilidade: 8
- 🧠 Inteligência: 4
```

## 📜 Grupo 2: História (Lore)
**Pasta de Trabalho:** `/historia`

Vocês são os roteiristas. Precisamos de um mundo! Cada membro pode criar um arquivo diferente para compor a lore (ex: `prologo.md`, `viloes.md`, `cidades.md`).

**💡 Dica de Estrutura para o Arquivo:**
```markdown
# Prólogo: A Queda do Servidor

Em um futuro distante, o grande Reino de [Nome do Mundo] era pacífico, até que...

### O Grande Vilão
**Nome:** [Nome do Vilão]
**Motivação:** Por que ele quer destruir ou dominar o mundo?

### Cidades e Locais Importantes
1. **A Taverna do Bug Infinito:** Local onde os heróis se encontram.
2. **A Caverna do Cache:** Um labirinto perigoso.
```

## 🐉 Grupo 3: Monstros e Batalha
**Pasta de Trabalho:** `/batalha`

Vocês cuidam da mecânica e dos inimigos! Dividam-se: alguns criam arquivos de regras (ex: `regras_combate.md`) e outros criam os inimigos (ex: `goblin.md`, `dragao_bugado.md`).

**💡 Dica de Estrutura para o Arquivo de Monstro:**
```markdown
# Dragão de Tela Azul

Um monstro aterrorizante que apaga a memória dos heróis.

### Status
- 💖 Vida (HP): 500
- 🛡️ Defesa: Alta

### Ataques
- **Fogo de Sintaxe:** Causa 30 de dano.
- **Crash do Sistema:** Paralisa o jogador por 1 turno.
```

**💡 Dica de Estrutura para Regras de Combate:**
```markdown
# Como Funciona a Batalha

1. O combate é em turnos.
2. Para atacar, jogue um Dado de 20 faces (D20).
3. Se tirar mais de 10, o ataque acerta!
```
